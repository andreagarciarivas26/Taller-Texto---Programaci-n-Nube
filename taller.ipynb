## Ejercicio 1. Cadena de funciones con prefijos

### Ejercicio 1 - parte 1

print_words_starting_with_a(path): leer las palabras del archivo path e imprimir las que empiezan por "a".

```python
path="words.txt"
def print_words_starting_with_a(path):
    for line in open(path):
        word = line.strip()
        if word.startswith("a"):
            print(word)
            
print_words_starting_with_a(path)            
```

### Ejercicio 1 parte 2
list_words_starting_with_a(path): el mismo filtro, pero retornar una lista (no imprimir).

```python
def list_words_starting_with_a(path):
    result = []                           #Se creo una lista vacía
    for line in open(path):
        word = line.strip()
        if word.startswith("a"):
            result.append(word)        #result.append guarda la palabra
    return result                            #Retorna la lista
print(list_words_starting_with_a(path))
```

### Ejercicio 1 parte 3
list_words_starting_with_an(path): retornar las palabras que empiezan por "an".

```python
def list_words_starting_with_an(path):
    result = []
    for line in open(path):
        word = line.strip()
        if word.startswith("an"):
            result.append(word)
    return result
print(list_words_starting_with_an(path))
```

### Ejercicio 1 parte 4
count_words_starting_with_an(path): retornar un diccionario {palabra: conteo} con las palabras que empiezan por "an".

```python
def count_words_starting_with_an(path):
    counter = {}                                              #Se crea el diccionario
    for word in list_words_starting_with_an(path):
        if word not in counter:
            counter[word] = 1
        else:
            counter[word] =counter[word] + 1
    return counter
print(count_words_starting_with_an(path))
```

### Ejercicio 2 - parte 1
lines_starting_with(path, prefix): retornar las líneas cuyo texto (sin espacios ni salto de línea) empieza por prefix.

```python
def lines_starting_with(history, prefix):
    result = []

    with open(history, "r", encoding="utf-8") as file:  
        for line in file:
            text = line.strip()

            if text.startswith(prefix):
                result.append(text)

    return result


print(lines_starting_with(history, "the"))
```

### Ejercicio 2 - parte 2

lines_ending_with(path, suffix): retornar las líneas que terminan en suffix (ignorando el salto de línea final).

```python
def lines_ending_with(history, sufijo):
    result = []

    with open(history, "r", encoding="utf-8") as file:  #¿Cúal es el significado utf-8?
        for line in file:
            text = line.strip()

            if text.endswith(sufijo):
                result.append(text)

    return result


print(lines_ending_with(history, "ing"))
```

### Ejercicio 2 - parte 3
lines_starting_or_ending(path, *affixes): retornar las líneas que empiezan o terminan con cualquiera de los prejijos *prefijos.

```python
def lines_starting_or_ending(history, *presuf):
    result = []

    with open(history, "r", encoding="utf-8") as file:
        for line in file:
            text = line.strip()
            if text.startswith(presuf) or text.endswith(presuf): #Te utiliza el operador or. Es True si alguno es True
                result.append(text)
        return result
print(lines_starting_or_ending(history,"of"))    
```

### Ejercicio 2 - parte 4
affix_histogram(path, *affixes): retornar un diccionario {afijo: n}, donde n es cuántas líneas empiezan o terminan con ese afijo. Una línea puede incrementar varias claves.

```python
def affix_histogram(history, *affixes):
    histogram={}   #Se crea un diccionario

    with open(history, "r", encoding="utf-8") as file:
            for line in file:
                text = line.strip()
                for affix in affixes:
                    if text.startswith(affix) or text.endswith(affix):
                        if affix not in histogram:
                            histogram[affix] = 1
                        else:
                            histogram[affix] =histogram[affix]+ 1
            return histogram
    
print(affix_histogram(history,"a","the"))
```

## Ejercicio 3  Catalog index (files + structured records)

#### Ejercicio 3 - item 1 
names_starting_with(path, prefix): devuelve los nombres de producto que empiezan con prefix.

```python
catalog="catalog.txt"
def names_starting_with(catalog, prefix):   #La función recibe dos parametros
    
   with open(catalog, "r", encoding="utf-8") as file:  # "r" es modo lectura, se le da el nombre de file
       
        names = []            #Crea una lista vacía
        for name in file:
            if name.startswith(prefix):
                names.append(name)
        return names  #Esta fuera del for y se ejecuta cuando se ha leido todo el archivo

print(names_starting_with(catalog,"an"))
```

#### Ejercicio 3 - item 2
count_by_name_prefix(path, prefix): devuelve un diccionario {nombre: cuántas veces aparece ese nombre} con los nombres que tienen ese prefijo.

```python
def count_by_name_prefix(path, prefix):
    counts = {}
    with open(path, "r", encoding="utf-8") as file:
        next(file)                                  # salta el encabezado
        for line in file:
            line = line.strip()                     # quita el '\n'
            if line == "":
                continue                              # ignora líneas vacías
            name, category, price = line.split("|")
            if name.startswith(prefix):
                counts[name] = counts.get(name, 0) + 1
    return counts

print(count_by_name_prefix("catalog.txt", "an"))
```

#### Ejercicio 3 - item 3
count_by_any_name_prefix(path, *prefixes): hace lo mismo, pero con varios prefijos (usando *prefixes y startswith con una tupla).

```python
def count_by_any_name_prefix(path, *prefixes):
    counts = {}
    with open(path, "r", encoding="utf-8") as file:
        next(file)                                  # salta el encabezado
        for line in file:
            line = line.strip()                     # quita el '\n'
            if line == "":
                continue                            # ignora líneas vacías
            name, category, price = line.split("|")
            if name.startswith(prefixes):           # prefixes es una tupla
                counts[name] = counts.get(name, 0) + 1
    return counts

print(count_by_any_name_prefix("catalog.txt", "an", "ap"))
```

#### Ejercicio 3 - item 4
group_names_by_prefix(path, *prefixes): devuelve {prefijo: [nombres que empiezan con él]}. Si un nombre coincide con más de un prefijo, va bajo el prefijo más largo. El estudiante debe decidir el orden: ordenar los prefijos por longitud, de mayor a menor.

```python
def group_names_by_prefix(path, *prefixes):
    groups = {pre: [] for pre in prefixes}              # una lista vacía por prefijo
    ordered = sorted(prefixes, key=len, reverse=True)   # del más largo al más corto
    with open(path, "r", encoding="utf-8") as file:
        next(file)                                  # salta el encabezado
        for line in file:
            line = line.strip()                     # quita el '\n'
            if line == "":
                continue                            # ignora líneas vacías
            name, category, price = line.split("|")
            for pre in ordered:
                if name.startswith(pre):
                    groups[pre].append(name)
                    break                           # se queda con el más largo
    return groups

print(group_names_by_prefix("catalog.txt", "a", "an", "and"))
```

## Exercise 4 — One-pass prefix report
Context
In the previous functions you either printed words, returned a list, or counted words for one prefix (or several prefixes, but only as “does this word match any of them?”).

Now you must read the file once and build a report that answers four questions together:

How many words are in the file?
How many of those words start with at least one of the given prefixes?
For each matched word, how many times does it appear?

```python
filename = "cuento.txt"
prefijo = "ed"

contador_total = 0
contador_palabras = {}

with open(filename, "r", encoding="utf-8") as file:   
    for line in file:
        palabras = line.split()

        for word in palabras:
            # 1. Contar todas las palabras del archivo
            contador_total = contador_total + 1

            # 2. Verificar si comienza con el prefijo
            if word.startswith(prefijo):

                # 3. Contar cuántas veces aparece cada palabra
                if word in contador_palabras:
                    contador_palabras[word] = contador_palabras[word] + 1
                else:
                    contador_palabras[word] = 1

print("Número total de palabras:", contador_total)

print("\nPalabras que comienzan con:", prefijo)

for word in contador_palabras:
    print(word, "→", contador_palabras[word])
```

```python
def prefix_report(path, *prefixes):
    total_words = 0                              # pregunta 1
    matched_words = 0                            # pregunta 2
    word_counts = {}                             # pregunta 3
    prefix_counts = {pre: 0 for pre in prefixes} # pregunta 4

    with open(path, "r", encoding="utf-8") as file:
        for line in file:                        # una sola lectura del archivo
            for word in line.split():
                total_words += 1                 # 1. cuenta todas las palabras

                if word.startswith(prefixes):    # ¿empieza con al menos un prefijo?
                    matched_words += 1           # cuenta las coincidencias
                    word_counts[word] = word_counts.get(word, 0) + 1   # 3. frecuencia

                    for pre in prefixes:         
                        if word.startswith(pre):
                            prefix_counts[pre] += 1

    return {
        "total_words": total_words,
        "matched_words": matched_words,
        "word_counts": word_counts,
        "prefix_counts": prefix_counts,
    }


report = prefix_report("cuento.txt", "ed", "edi", "wa")

print("Total de palabras:", report["total_words"])
print("Palabras que empiezan con algún prefijo:", report["matched_words"])

print("\nFrecuencia de cada palabra encontrada:")
for word in report["word_counts"]:
    print(" ", word, "→", report["word_counts"][word])

print("\nPalabras por prefijo:")
for pre in report["prefix_counts"]:
    print(" ", pre, "→", report["prefix_counts"][pre])
```
